```python
#Loading dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from sklearn.decomposition import PCA

sns.set(style="whitegrid")

train_path = r"C:\Users\kbc_k\comp3000\UNSW_NB15_training-set1.csv"
test_path  = r"C:\Users\kbc_k\comp3000\UNSW_NB15_testing-set2.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print(train_df.head())
```

```python
#lavel count
print("Train columns:")
print(train_df.columns)

print("\nTrain label counts:")
print(train_df["label"].value_counts())

print("\nTest label counts:")
print(test_df["label"].value_counts())

print("\nTrain info:")
print(train_df.info())
# check how many 0 nad 1 labels are there? 0 =normal and 1= attack
```

```python
#missing value in train
print("Missing values in train:")
print(train_df.isnull().sum())

print("\nMissing values in test:")
print(test_df.isnull().sum())

train_df = train_df.replace([np.inf, -np.inf], np.nan)
test_df = test_df.replace([np.inf, -np.inf], np.nan)

train_df = train_df.dropna()
test_df = test_df.dropna()

print("Train shape after cleaning:", train_df.shape)
print("Test shape after cleaning:", test_df.shape)
```

```python
#features selection and remove unnecessary columns
label_col = "label"
drop_cols = ["label", "attack_cat", "id"]

X_train = train_df.drop(columns=drop_cols, errors="ignore")
y_train = train_df[label_col]

X_test = test_df.drop(columns=drop_cols, errors="ignore")
y_test = test_df[label_col]

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
#identifying cat and numeric columns
categorical_cols = X_train.select_dtypes(include=["object"]).columns.tolist()
numeric_cols = X_train.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical columns:", categorical_cols)
print("Numeric columns count:", len(numeric_cols))
```

```python
#preprocess data
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", StandardScaler(), numeric_cols)
    ]
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed train shape:", X_train_processed.shape)
print("Processed test shape:", X_test_processed.shape)
```

```python
#train Naive Bayes Model 

from sklearn.preprocessing import MinMaxScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Use numeric-only version for Naive Bayes
X_train_nb = X_train[numeric_cols].copy()
X_test_nb = X_test[numeric_cols].copy()

X_train_nb = X_train_nb.replace([np.inf, -np.inf], np.nan).fillna(0)
X_test_nb = X_test_nb.replace([np.inf, -np.inf], np.nan).fillna(0)

scaler_nb = MinMaxScaler()
X_train_nb_scaled = scaler_nb.fit_transform(X_train_nb)
X_test_nb_scaled = scaler_nb.transform(X_test_nb)

nb_model = GaussianNB(var_smoothing=1e-6)
nb_model.fit(X_train_nb_scaled, y_train)

nb_preds = nb_model.predict(X_test_nb_scaled)

print("Naive Bayes Results")
print("Accuracy:", accuracy_score(y_test, nb_preds))
print("Precision:", precision_score(y_test, nb_preds, zero_division=0))
print("Recall:", recall_score(y_test, nb_preds, zero_division=0))
print("F1-score:", f1_score(y_test, nb_preds, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, nb_preds))

print("\nClassification Report:")
print(classification_report(y_test, nb_preds, zero_division=0))

import matplotlib.pyplot as plt
def plot_pie(y_true, y_pred, title):
    correct = (y_true == y_pred).sum()
    incorrect = (y_true != y_pred).sum()

    labels = ["Correct", "Incorrect"]
    sizes = [correct, incorrect]

    plt.figure(figsize=(5,5))
    plt.pie(sizes, labels=labels, autopct='%1.1f%%',
            colors=["green", "red"], startangle=90)
    plt.title(title)
    plt.show()

# ---- call function for correct and incorrect prediction
plot_pie(y_test, nb_preds, "Naive Bayes Prediction Accuracy")
```

```python
#Train KNN
from sklearn.neighbors import KNeighborsClassifier
from sklearn.utils import resample

# Smaller subset for KNN speed
train_sample_size = 30000
test_sample_size = 10000

train_small = train_df.sample(n=min(train_sample_size, len(train_df)), random_state=42)
test_small = test_df.sample(n=min(test_sample_size, len(test_df)), random_state=42)

X_train_knn = train_small.drop(columns=drop_cols, errors="ignore")
y_train_knn = train_small[label_col]

X_test_knn = test_small.drop(columns=drop_cols, errors="ignore")
y_test_knn = test_small[label_col]

categorical_cols_knn = X_train_knn.select_dtypes(include=["object"]).columns.tolist()
numeric_cols_knn = X_train_knn.select_dtypes(exclude=["object"]).columns.tolist()

preprocessor_knn = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols_knn),
        ("num", StandardScaler(), numeric_cols_knn)
    ]
)

X_train_knn_processed = preprocessor_knn.fit_transform(X_train_knn)
X_test_knn_processed = preprocessor_knn.transform(X_test_knn)

knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_knn_processed, y_train_knn)

knn_preds = knn_model.predict(X_test_knn_processed)

print("KNN Results")
print("Accuracy:", accuracy_score(y_test_knn, knn_preds))
print("Precision:", precision_score(y_test_knn, knn_preds, zero_division=0))
print("Recall:", recall_score(y_test_knn, knn_preds, zero_division=0))
print("F1-score:", f1_score(y_test_knn, knn_preds, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_knn, knn_preds))

print("\nClassification Report:")
print(classification_report(y_test_knn, knn_preds, zero_division=0))
import matplotlib.pyplot as plt
def plot_pie(y_true, y_pred, title):
    correct = (y_true == y_pred).sum()
    incorrect = (y_true != y_pred).sum()

    labels = ["Correct", "Incorrect"]
    sizes = [correct, incorrect]

    plt.figure(figsize=(5,5))
    plt.pie(sizes, labels=labels, autopct='%1.1f%%',
            colors=["green", "red"], startangle=90)
    plt.title(title)
    plt.show()

# ---- call function for correct and incorrect prediction
plot_pie(y_test_knn, knn_preds, "KNN Prediction Accuracy")
```

```python
#Train SVM
from sklearn.svm import LinearSVC

svm_model = LinearSVC(
    C=1.0,
    class_weight="balanced",
    max_iter=8000,
    random_state=42
)

svm_model.fit(X_train_processed, y_train)

svm_preds = svm_model.predict(X_test_processed)

print("SVM Results")
print("Accuracy:", accuracy_score(y_test, svm_preds))
print("Precision:", precision_score(y_test, svm_preds, zero_division=0))
print("Recall:", recall_score(y_test, svm_preds, zero_division=0))
print("F1-score:", f1_score(y_test, svm_preds, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, svm_preds))

print("\nClassification Report:")
print(classification_report(y_test, svm_preds, zero_division=0))

import matplotlib.pyplot as plt
def plot_pie(y_true, y_pred, title):
    correct = (y_true == y_pred).sum()
    incorrect = (y_true != y_pred).sum()

    labels = ["Correct", "Incorrect"]
    sizes = [correct, incorrect]

    plt.figure(figsize=(5,5))
    plt.pie(sizes, labels=labels, autopct='%1.1f%%',
            colors=["green", "red"], startangle=90)
    plt.title(title)
    plt.show()

# ---- call function for correct and incorrect prediction
plot_pie(y_test, svm_preds, "SVM Prediction Accuracy")


```

```python
#Train Random Forest
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_processed, y_train)

rf_preds = rf_model.predict(X_test_processed)

print("Random Forest Results")
print("Accuracy:", accuracy_score(y_test, rf_preds))
print("Precision:", precision_score(y_test, rf_preds, zero_division=0))
print("Recall:", recall_score(y_test, rf_preds, zero_division=0))
print("F1-score:", f1_score(y_test, rf_preds, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, rf_preds))

print("\nClassification Report:")
print(classification_report(y_test, rf_preds, zero_division=0))

import matplotlib.pyplot as plt
def plot_pie(y_true, y_pred, title):
    correct = (y_true == y_pred).sum()
    incorrect = (y_true != y_pred).sum()

    labels = ["Correct", "Incorrect"]
    sizes = [correct, incorrect]

    plt.figure(figsize=(5,5))
    plt.pie(sizes, labels=labels, autopct='%1.1f%%',
            colors=["green", "red"], startangle=90)
    plt.title(title)
    plt.show()

# ---- call function for correct and incorrect prediction
plot_pie(y_test, rf_preds, "Random Forest Prediction Accuracy")
```

```python
#Result of all four algorithms
results_unsw = pd.DataFrame({
    "Algorithm": ["Naive Bayes", "KNN", "SVM", "Random Forest"],
    "Accuracy": [
        accuracy_score(y_test, nb_preds),
        accuracy_score(y_test_knn, knn_preds),
        accuracy_score(y_test, svm_preds),
        accuracy_score(y_test, rf_preds)
    ],
    "Precision": [
        precision_score(y_test, nb_preds, zero_division=0),
        precision_score(y_test_knn, knn_preds, zero_division=0),
        precision_score(y_test, svm_preds, zero_division=0),
        precision_score(y_test, rf_preds, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test, nb_preds, zero_division=0),
        recall_score(y_test_knn, knn_preds, zero_division=0),
        recall_score(y_test, svm_preds, zero_division=0),
        recall_score(y_test, rf_preds, zero_division=0)
    ],
    "F1-score": [
        f1_score(y_test, nb_preds, zero_division=0),
        f1_score(y_test_knn, knn_preds, zero_division=0),
        f1_score(y_test, svm_preds, zero_division=0),
        f1_score(y_test, rf_preds, zero_division=0)
    ]
})

print("UNSW-NB15 Results Table")
print(results_unsw)
```

```python
#plot PCA visualisation
numeric_features_only = train_df.select_dtypes(exclude=["object"]).drop(columns=["label"], errors="ignore")
labels = train_df["label"]

pca_sample_X = numeric_features_only.sample(n=min(5000, len(numeric_features_only)), random_state=42)
pca_sample_y = labels.loc[pca_sample_X.index]

scaler_pca = StandardScaler()
X_scaled_pca = scaler_pca.fit_transform(pca_sample_X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled_pca)

plt.figure(figsize=(8,6))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=pca_sample_y, palette="coolwarm", alpha=0.7)
plt.title("PCA Visualization of Normal vs Attack Traffic")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Label")
plt.show()
```
